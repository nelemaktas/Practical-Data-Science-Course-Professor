# Data Visualization I: Telling Stories with Plotly

This notebook demonstrates how to use Plotly for data visualization, covering basic and advanced plots, customization, interactivity, storytelling, and best practices.

## 1. Import Required Libraries
We'll use pandas for data handling and plotly.express for visualization.

```python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go
```

## 2. Load Example Dataset
We'll load the average monthly surface temperature dataset from the data folder.

```python
df = pd.read_csv('../data/average-monthly-surface-temperature.csv')
# Convert 'Day' column to datetime
df['Day'] = pd.to_datetime(df['Day'])
df['Year'] = df['Day'].dt.year
df_2007 = df[df['Year'] == 2007]  # Focus on 2007 for clarity
print(f"Dataset shape: {df.shape}")
print(f"Dataset shape for 2007: {df_2007.shape}")
df.head()
```

```python
df_2007.describe()
```

```python
df_2007.nunique()
```

## 3. Bar Plot: Average Temperature by Country
Let's visualize average monthly temperature by country for 2007.

```python
# Get average temperature by country for 2007
temp_by_entity = df_2007.groupby('Entity')['Monthly average'].mean().reset_index()
temp_by_entity.columns = ['Entity', 'AvgTemp']
temp_by_entity = temp_by_entity.sort_values('AvgTemp', ascending=False).head(20)
fig = px.bar(temp_by_entity, x='AvgTemp', y='Entity',
             title='Top 20 Countries by Average Temperature (2007)',
             labels={'AvgTemp': 'Average Temperature (°C)', 'Entity': 'Country'})
fig.show()
```

## 4. Bar Plot: Temperature Range by Country
A bar plot to compare temperature ranges across different countries.

```python
temp_range = df_2007.groupby('Entity')['Monthly average'].agg(['min', 'max']).reset_index() # for each entity, gives min and max of the monthly average in 2007
temp_range['range'] = temp_range['max'] - temp_range['min'] # add column for range
temp_range = temp_range.sort_values('range', ascending=False).head(15) # restrict to the top 15 countries
fig = px.bar(temp_range, x='Entity', y='range',
             title='Temperature Range by Country (2007)',
             labels={'range': 'Temperature Range (°C)', 'Entity': 'Country'})
fig.show()
```

## 5. Line Plot: Temperature Over Time
Let's see how temperature has changed over time for a specific country.

```python
fig = px.line(df[df['Entity'] == 'France'], x='Day', y='Monthly average',
              title='Monthly Average Temperature in France Over Time',
              labels={'Monthly average': 'Temperature (°C)', 'Day': 'Date'})
fig.show()
```

## 6. Customizing Visualizations
You can customize colors, labels, and layout for clarity and impact.

```python
# Compare monthly temperature trends for selected countries in 2007
selected_countries = ['Canada', 'Brazil', 'Egypt', 'Russia']
df_selected = df_2007[df_2007['Entity'].isin(selected_countries)] # Df_selected filters df_2007 for those countries that are in the list "selected_countries"
fig = px.line(df_selected, x='Day', y='Monthly average',
              color='Entity',
              title='Customized: Monthly Temperature Trends for Selected Countries (2007)',
              labels={'Monthly average': 'Temperature (°C)', 'Day': 'Date'})
fig.update_layout(legend_title_text='Country',
                  plot_bgcolor='rgba(240,240,240,0.95)')
fig.show()
```

## 7. Interactive Charts
Plotly charts are interactive by default: try zooming, hovering, and filtering.

Let's create an interactive chart showing temperature distribution.

```python
# Create an interactive box plot showing temperature distribution by top countries
## .index returns just the names (labels) of those top entities.
top_entities = df_2007.groupby('Entity')['Monthly average'].mean().nlargest(8).index
df_top = df_2007[df_2007['Entity'].isin(top_entities)]

fig = px.box(df_top, x='Entity', y='Monthly average',
             title='Temperature Distribution by Top 8 Countries (2007)',
             labels={'Monthly average': 'Temperature (°C)'})

fig.update_layout(
    hovermode='x unified',
    showlegend=True
)

fig.show()
```

## 1) Compare monthly average temperature per continent in a boxplot

```python
df_continents = df[df['Entity'].isin(continents)]

# fig = px.box(df_continents, x='Entity', y='Monthly average', title='Temperature Distribution by Continents', labels={'Monthly average': 'Temperature (°C)'})

order = (df_continents.groupby('Entity')['Monthly average'].mean().sort_values().index)

fig = px.box(df_continents,
    x='Entity',
    y='Monthly average',
    category_orders={'Entity': order},
    title='Temperature Distribution by Continents',
    labels={'Monthly average': 'Temperature (°C)'}
)

fig.update_layout(
    hovermode='x unified',
    showlegend=True
)

fig.show()
```

```python
df_oceans = df[df['Entity'].isin(oceans)]

order = (df_oceans.groupby('Entity')['Monthly average'].mean().sort_values().index)

fig = px.box(df_oceans, x='Entity', y='Monthly average', category_orders={'Entity': order},
             title='Temperature Distribution of the Oceans',
             labels={'Monthly average': 'Temperature (°C)'})
fig.show()
```

## Compare the yearly average temperature of continental and coastal regions in different continents in a line graph

```python
df_oceans_yearly = df_oceans.groupby(['Entity', df_oceans['Day'].dt.year])['Monthly average'].mean().reset_index()
df_oceans_yearly.columns = ['Entity','Year', 'YearlyAvgTemp']
df_oceans_yearly.head()
```

```python

fig = px.line(df_continents_yearly[df_continents_yearly['Year']>2000], x='Year', y='YearlyAvgTemp', color='Entity', title='Yearly Temperature Trends for Continents')
fig.show()
```

```python
fig = px.line(df_oceans_yearly[df_oceans_yearly['Year']>2000], x='Year', y='YearlyAvgTemp', color='Entity', title='Yearly Temperature Trends for Oceans')
fig.show()
```

Idea: bar chart for yearly temperature range (max - min) for each continent and ocean in one chart ---- OR: only for North America - North Atlantic Ocean - North Pacific Ocean

```python
df_nordic = df[df['Entity'].isin(['Antarctica (NIAID)', 'Arctic Ocean (NIAID)', 'Southern Ocean (NIAID)'])]
fig = px.line(df_nordic, x='Year', y='Monthly average', color='Entity', title='Yearly Temperature Trends for Arctic, Antarctic, and Southern Oceans based on monthly data')
fig.show()
```

```python
# Compute yearly average per entity
yearly_avg = (
    df_nordic
    .groupby(['Entity', 'Year'])['Monthly average']
    .mean()
    .reset_index()
)

# Filter years 2000–2025
yearly_avg = yearly_avg[
    (yearly_avg['Year'] >= 2000) &
    (yearly_avg['Year'] <= 2025)]

fig = px.line(
    yearly_avg,
    x='Year',
    y='Monthly average',
    color='Entity',
    title='Yearly Temperature Trends for Arctic, Antarctic, and Southern Oceans'
)

fig.show()
```

```python
# Keep only 2000 and 2025
selected = yearly_avg[yearly_avg['Year'].isin([2000, 2025])]

# Sort to ensure 2000 comes before 2025
selected = selected.sort_values(['Entity', 'Year'])

# Compute difference (2025 - 2000)
selected['Change'] = selected.groupby('Entity')['Monthly average'].diff()

# Keep only 2025 rows (these now contain the difference)
result = selected[selected['Year'] == 2025]

# Plot bar chart
fig = px.bar(
    result,
    x='Entity',
    y='Change',
    title='Temperature Change (2025 vs 2000)',
    labels={'Change': 'Temperature Change (°C)'}
)

fig.show()
```
