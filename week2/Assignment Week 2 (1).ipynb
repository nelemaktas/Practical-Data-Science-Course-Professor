```python

import pandas as pd
df = pd.read_csv("https://ourworldindata.org/grapher/mean-years-of-schooling-long-run.csv?v=1&csvType=full&useColumnShortNames=true", storage_options = {'User-Agent': 'Our World In Data data fetch/1.0'})
print(f"Loaded dataset with {df.shape[0]:,} rows and {df.shape[1]} columns.")
```

```python
df.isnull().sum().sort_values(ascending=False).head(10)

#df['missing_count'] = df.isnull().sum(axis=1)
#df.sort_values('missing_count', ascending=False).head(10)
```

```python
df_clean = df.dropna(subset=['mf_youth_and_adults__15_64_years__average_years_of_education'])
df_clean = df_clean.rename(columns={'entity':'country', 'mf_youth_and_adults__15_64_years__average_years_of_education':'mean_years_schooling'})
df_clean.head()
```

```python
df_clean.shape
```

```python
df['entity'] = df['entity'].replace('Turkey','Türkiye')
#df[df['entity']=='Türkiye']
```

```python
top_schooling = df_clean[df_clean['year'] == 2020].sort_values('mean_years_schooling', ascending=False)
top_schooling[['country','mean_years_schooling']].head()
```

```python
low_schooling = df_clean[df_clean['year']== 2020].sort_values('mean_years_schooling', ascending=True)
low_schooling[['country','mean_years_schooling']].head(10)
```

```python
country_avg = df_clean.groupby('country')['mean_years_schooling'].mean().sort_values(ascending=False)
country_avg.head()
```

```python
rank = country_avg.index.get_loc('Germany')+1
print("Germany is ranked:", rank)
```

```python
# What is the average mean years of schooling for the top 10 countries in 2020?
# IDEA create set that only contains top 10 countries. find avg of those in 2018
schooling_2018_sorted = df_clean[df_clean['year']== 2020].sort_values('mean_years_schooling', ascending=False)
top_ten_in_2018 = schooling_2018_sorted.head(10).reset_index()
print("The average mean years of schooling for the top 10 countries in 2018 is", top_ten_in_2018['mean_years_schooling'].mean())

```

```python
# Which countries have shown the largest increase in mean years of schooling from the earliest to the latest year?
# IDEA new column for the difference between earliest and latest year, then sort
df_sorted = df_clean.sort_values('year')
schooling_change = (df_clean.groupby('country')['mean_years_schooling'].last() - df_clean.groupby('country')['mean_years_schooling'].first())
schooling_change.sort_values(ascending=False).head(10)
```

```python
# How has mean years of schooling changed over time for Germany?
# IDEA fetch first year of Germany, fetch last year, find average for every century, find total average, plot all
df_Germany = df_clean[df_clean['country'] == 'Germany']

first_year_value = df_Germany.iloc[0]
last_year_value = df_Germany.iloc[-1]

country_avg.loc["Germany"]
country_avg_Germany = df_Germany.groupby('year')['mean_years_schooling'].mean()
country_avg_Germany.plot(title="Germany's Mean Years of Schooling Over Time")
print("In the first year, which was", first_year_value['year'], "Germany had a mean years of schooling of", first_year_value['mean_years_schooling'])
print("In the last year, which was", last_year_value['year'], "Germany had a mean years of schooling of", last_year_value['mean_years_schooling'])
print("The average mean years of schooling for Germany is", country_avg.loc["Germany"])
df_clean["century"] = ((df_clean["year"]-1)//100)+1
#df_Germany.groupby('century')['mean_years_schooling'].mean().plot(title="Germany's Mean Years of Schooling Over Time")
```
